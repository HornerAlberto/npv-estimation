# NPV estimation

Uses `npv.py`. Cash flows are one per period, starting at t=0 (not discounted).
The rate is **annual** and is converted to the period length with effective compounding:

$$r_p = (1 + r_{annual})^{1/n} - 1, \qquad NPV = \sum_{t=0}^{T} \frac{CF_t}{(1 + r_p)^t}$$

| period | n per year |
|---|---|
| `year` | 1 |
| `semester` | 2 |
| `week` | 52 |

```python
import pandas as pd

from npv import npv, npv_table, periodic_rate
```

## Inputs

```python
annual_rate = 0.10                       # 10% effective annual
cash_flows = [-1000, 300, 400, 500]      # CF_0 is the initial investment
period = "year"                          # "year", "semester" or "week"
```

## Result

```python
print(f"Rate per {period}: {periodic_rate(annual_rate, period):.6%}")
print(f"NPV: {npv(annual_rate, cash_flows, period):,.2f}")
```

```python
pd.DataFrame(npv_table(annual_rate, cash_flows, period)).set_index("period")
```

## Same flows, different period lengths

The same list of cash flows means different timing depending on `period`: shorter periods arrive sooner, so they are discounted less.

```python
pd.DataFrame(
    {
        p: {
            "rate per period": periodic_rate(annual_rate, p),
            "NPV": npv(annual_rate, cash_flows, p),
        }
        for p in ["year", "semester", "week"]
    }
).T
```
